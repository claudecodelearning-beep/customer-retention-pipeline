# Data audit

One row per customer snapshot in the public IBM Telco sample. The 7,043 rows have unique customer IDs and no fully duplicated rows. Churn is the recorded target; 1,869 customers are labeled Yes (26.54%). TotalCharges is stored as text and contains 11 blank values. These are converted to missing numeric values and imputed only inside training folds. Customer IDs are kept for row alignment, never used as predictors. This static dataset supports association and offline classification, not causal effects, future timing or measured retention ROI.
